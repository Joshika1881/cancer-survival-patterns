```python
import pandas as pd

df = pd.read_csv("../data/seer_breast_cancer.csv")

df.head()
```

```python
# Remove the empty column
df = df.drop(columns=["Unnamed: 3"])

df.shape
```

```python
# Check for duplicate rows
df.duplicated().sum()
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

df.shape
```

```python
# Remove extra spaces from column names
df.columns = df.columns.str.strip()

df.columns.tolist()
```

```python
# Fix misspelled column name
df = df.rename(columns={
    "Reginol Node Positive": "Regional Node Positive"
})

df.columns.tolist()
```

```python
print("Cleaned dataset shape:", df.shape)
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
print("\nColumn names:")
print(df.columns.tolist())
```

```python
# Save cleaned dataset
df.to_csv("../data/seer_breast_cancer_cleaned.csv", index=False)
```

```python
df = df.drop(columns=["Unnamed: 3"])
```

```python
df = df.drop_duplicates()
```

```python
df.columns = df.columns.str.strip()
```

```python
df = df.rename(columns={
    "Reginol Node Positive": "Regional Node Positive"
})
```

```python
print("Shape:", df.shape)
print("Duplicates:", df.duplicated().sum())
print("Missing values:", df.isnull().sum().sum())
print("\nColumns:")
print(df.columns.tolist())
```

```python
df["Death"] = df["Status"].map({
    "Alive": 0,
    "Dead": 1
})

df[["Status", "Death"]].value_counts()
```

```python
df["Age Group"] = pd.cut(
    df["Age"],
    bins=[0, 49, 59, 69],
    labels=["Under 50", "50-59", "60-69"]
)

df["Age Group"].value_counts().sort_index()
```

```python
df["Hormone Status"] = (
    (df["Estrogen Status"] == "Positive") |
    (df["Progesterone Status"] == "Positive")
).map({
    True: "Positive",
    False: "Negative"
})

df["Hormone Status"].value_counts()
```

```python
df["Grade Group"] = df["Grade"].map({
    "Well differentiated; Grade I": "Grade I",
    "Moderately differentiated; Grade II": "Grade II",
    "Poorly differentiated; Grade III": "Grade III-IV",
    "Undifferentiated; anaplastic; Grade IV": "Grade III-IV"
})

df["Grade Group"].value_counts()
```

```python
df["Stage Group"] = df["6th Stage"].map({
    "IIA": "Stage II",
    "IIB": "Stage II",
    "IIIA": "Stage III",
    "IIIB": "Stage III",
    "IIIC": "Stage III"
})

df["Stage Group"].value_counts()
```

```python
print("Dataset shape:", df.shape)

print("\nDeath:")
print(df["Death"].value_counts())

print("\nAge Group:")
print(df["Age Group"].value_counts().sort_index())

print("\nHormone Status:")
print(df["Hormone Status"].value_counts())

print("\nGrade Group:")
print(df["Grade Group"].value_counts())

print("\nStage Group:")
print(df["Stage Group"].value_counts())

print("\nMissing values in new variables:")
print(
    df[
        ["Death", "Age Group", "Hormone Status", "Grade Group", "Stage Group"]
    ].isnull().sum()
)
```

```python
df.to_csv("../data/seer_breast_cancer_cleaned.csv", index=False)

print("Updated cleaned dataset saved.")
```
